```python
import os
from pathlib import Path
import pandas as pd
```

```python
df_raw = pd.read_csv('../data/joined_data/olist_joined_dataset.csv')
```

```python
def profile_data(df: pd.DataFrame) -> pd.DataFrame:
    """In kích thước, kiểu dữ liệu và bảng missing values của dữ liệu thô.

    Mục đích: có "bức tranh gốc" để đối chiếu với Row Audit sau này
    (biết mỗi bước làm sạch đã loại bao nhiêu dòng so với ban đầu).
    """
    print("=" * 70)
    print("DATA PROFILING")
    print("=" * 70)

    # shape = (số dòng, số cột)
    print(f"Shape: {df.shape[0]:,} dòng x {df.shape[1]} cột\n")

    # dtypes cho biết cột nào đang bị đọc sai kiểu
    # (ví dụ timestamp đang là chuỗi 'object' thay vì datetime)
    print("Kiểu dữ liệu (dtypes):")
    print(df.dtypes.to_string(), "\n")

    missing = pd.DataFrame(
        {
            # isna().sum(): đếm số ô trống theo từng cột
            "missing_count": df.isna().sum(),
            # isna().mean(): tỷ lệ ô trống (0-1), nhân 100 để ra phần trăm
            "missing_pct": (df.isna().mean() * 100).round(2),
        }
    ).sort_values("missing_count", ascending=False)  # cột thiếu nhiều nhất lên đầu
    print("Missing values theo cột:")
    print(missing.to_string(), "\n")
    return missing
```

```python
profile_df_raw = profile_data(df_raw)
```
